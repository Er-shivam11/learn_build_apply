# 🟢 Problem: Two Sum
## 📌 Question

Given an array of numbers `nums` and a number `target`, find **two different numbers** in the array whose sum equals the target.

Return the **indexes** of those two numbers.

---

## 📥 Input/output

```python
nums = [2, 7, 11, 15]
target = 9
output = [0, 1]


## 🧠 Step-by-Step Thinking

**Step 1 → Read the question once.**

**Step 2 → What do they want?**  
→ Find **TWO numbers**.

**Step 3 → What should those two numbers do?**  
→ Their **sum = target**.

**Step 4 → What should I return?**  
→ Their **INDEXES**, not the numbers.

**Step 5 → Since I need BOTH index and value:**  
→ Use `enumerate()`.

`for i, num in enumerate(nums):`

**Step 6 → Can I check every pair?**  
→ Yes → Brute Force → but it is slow **O(n²)**.

**Step 7 → Can I do it in one pass?**  
→ Yes.

**Step 8 → Have I already seen the number I need?**  
→ Use a **HashMap** because it is like a small notebook where I can quickly check whether a number has already appeared.

**Step 9 → What number do I need?**

`complement = target - num`

**Step 10 → Is the complement already in my notebook?**  
→ YES → I found the answer.

**Step 11 → If NOT:**  
→ Store the current number and its index.

**Step 12 → Move to the next number.**

**Step 13 → Repeat until the answer is found.**

---

## 🔍 Recognition

✓ **Array**  
→ Input is a list of numbers.

✓ **Traversal**  
→ I need to look through the numbers.

✓ **HashMap**  
→ I need to remember previous numbers and find them quickly.

✓ **enumerate()**  
→ I need both **index + value**.

---

````python
print("hello ")
````

````python
nums = [2,7,11,15]
target = 9
seen = {}
for idx,num in enumerate(nums):
    compliment=target-num
    if compliment in seen:
        seen[compliment],num
    seen[num]=idx
````

````python
class Solution(object):
    def twoSum(self, nums, target):
        seen = {}

        for i, num in enumerate(nums):
            complement = target - num

            if complement in seen:
                return [seen[complement], i]

            seen[num] = i
````

# 🟢 Problem: Maximum Subarray

## 📌 Question

Given an integer array `nums`, find the continuous subarray with the largest sum and return its sum.

A subarray means a continuous part of the array.

---

## 📥 Input / Output

Input:
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

Output:
6

Why?

The subarray with the largest sum is:

[4, -1, 2, 1]

4 + (-1) + 2 + 1 = 6

---

## 🧠 Step-by-Step Thinking

Step 1 → Read the question once.

Step 2 → What do they want?
→ Find a SUBARRAY.

Step 3 → What does subarray mean?
→ A CONTINUOUS part of the array.

Step 4 → What should I return?
→ The LARGEST SUM, not the subarray.

Step 5 → Do I need the indexes?
→ No.

Step 6 → Do I need both index and value?
→ No → I only need the VALUE.

Step 7 → Which loop should I use?
→ `for num in nums:`

Why?
→ I only need each number.

Step 8 → Can I check every possible subarray?
→ Yes → Brute Force, but it is slow.

Step 9 → Can I solve it in one traversal?
→ Yes.

Step 10 → Ask:
"Should I continue my current subarray or start a new one?"

Step 11 → Start fresh:
→ `current = num`

Step 12 → Continue previous subarray:
→ `current = current + num`

Combine both decisions:

→ `current = max(num, current + num)`

Step 13 → What else do I need to remember?
→ The BEST sum found so far.

→ `best = max(best, current)`

Step 14 → Move to the next number.

Step 15 → Repeat until the array ends.

Step 16 → Return `best`.

---

## 🔍 Recognition

✓ Array
→ Input is a list of numbers.

✓ Subarray
→ We need a CONTINUOUS part of the array.

✓ Traversal
→ We look at every number once.

✓ Value Traversal
→ We only need the current number, not its index.

✓ Kadane's Algorithm
→ At every number, decide CONTINUE or RESTART.

---

## 🧩 Templates Loaded

Value Traversal:
`for num in nums:`

Current Subarray:
`current = max(num, current + num)`

Best Sum:
`best = max(best, current)`

---

---

## 🧠 Final Memory Shortcut

SUBARRAY
↓
CONTINUOUS
↓
Value Traversal
↓
Current Sum
↓
CONTINUE or RESTART
↓
`current = max(num, current + num)`
↓
Remember Best
↓
`best = max(best, current)`


|  Step | `num` | `current + num` | `max(num, current + num)` → `current` |`best=max(best, current) `|
| ----: | ----: | --------------: | ------------------------------------: | ------------------: |
| Start |     — |               — |                        `current = -2` |         `best = -2` |
|     1 |   `1` |   `-2 + 1 = -1` |                      `max(1, -1) = 1` |    `max(-2, 1) = 1` |
|     2 |  `-3` |   `1 + -3 = -2` |                    `max(-3, -2) = -2` |    `max(1, -2) = 1` |
|     3 |   `4` |    `-2 + 4 = 2` |                       `max(4, 2) = 4` |     `max(1, 4) = 4` |
|     4 |  `-1` |    `4 + -1 = 3` |                      `max(-1, 3) = 3` |     `max(4, 3) = 4` |
|     5 |   `2` |     `3 + 2 = 5` |                       `max(2, 5) = 5` |     `max(4, 5) = 5` |
|     6 |   `1` |     `5 + 1 = 6` |                       `max(1, 6) = 6` |     `max(5, 6) = 6` |
|     7 |  `-5` |    `6 + -5 = 1` |                      `max(-5, 1) = 1` |     `max(6, 1) = 6` |
|     8 |   `4` |     `1 + 4 = 5` |                       `max(4, 5) = 5` |     `max(6, 5) = 6` |


````python
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current = nums[0]
best = nums[0]

for num in nums[1:]:
        current=max(num,current+num)
        best=max(best, current)
print(best)
````

# 🟢 Problem: Contains Duplicate

## 📌 Question

Given an integer array `nums`, return `True` if any value appears **more than once** in the array.

Otherwise, return `False`.

---

## 📥 Input / Output

Input:

nums = [1, 2, 3, 1]

Output:

True

Because:

1 appears more than once.

---

## 🧠 Step-by-Step Thinking

**Step 1 → What does the problem want?**

→ Check whether any number appears more than once.

**Step 2 → What does duplicate mean?**

→ The same number appears again in the array.

**Step 3 → What should I return?**

→ `True` or `False`.

**Step 4 → Do I need indexes?**

→ No.

**Step 5 → Do I need to remember anything?**

→ Yes, I need to remember the numbers I have already seen.

**Step 6 → What data structure should I use?**

→ HashSet / `set()`.

**Why Set?**

→ Because I only need to ask:

"Have I seen this number before?"

I don't need its index or any extra information.

**Step 7 → Which loop should I use?**

→ Normal value traversal.

`for num in nums:`

**Step 8 → What question do I ask for every number?**

→ Is `num` already inside `seen`?

**Step 9 → If YES?**

→ Duplicate found → return `True`.

**Step 10 → If NO?**

→ Add this new number to `seen`.

**Step 11 → Continue until the array ends.**

**Step 12 → If no duplicate was found?**

→ Return `False`.

---

## 🔍 Recognition

✓ **Array**

→ Input is a list of numbers.

✓ **Traversal**

→ I need to check every number.

✓ **HashSet**

→ I only need to remember whether a number was seen before.

✓ **No `enumerate()`**

→ I don't need indexes.

---




## 🧪 Local Test

nums = [1, 2, 3, 1]

solution = Solution()

print(solution.containsDuplicate(nums))


## 🔄 Execution Flow

num = 1
seen = {}
→ 1 not found
→ add 1

seen = {1}

num = 2
→ 2 not found
→ add 2

seen = {1, 2}

num = 3
→ 3 not found
→ add 3

seen = {1, 2, 3}

num = 1
→ 1 already exists

return True


## 🧠 Final Logic

Current number
↓
Have I seen it before?
↓
YES → return True
↓
NO → add it to seen
↓
Move to next number
↓
No duplicates → return False

````python
## 💻 Code

def containsDuplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False
nums=[1,2,3,2,4,4,5]
print(containsDuplicate(nums))
````

# 🟢 Problem: Best Time to Buy and Sell Stock

## 📌 Question

You are given an array `prices` where:

`prices[i]` = price of a stock on day `i`.

Choose **one day to buy** and a **later day to sell**.

Return the **maximum profit**.

You cannot sell before you buy.

---

## 📥 Input / Output

Input:

prices = [7, 1, 5, 3, 6, 4]

Output:

5

Because:

Buy at `1`

Sell at `6`

Profit:

6 - 1 = 5

---

## 🧠 Step-by-Step Thinking

**Step 1 → What does the problem want?**

→ Find the **maximum profit**.

**Step 2 → How do we calculate profit?**

→

profit = selling_price - buying_price

**Step 3 → What must happen first?**

→ Buy first, then sell later.

**Step 4 → What should I return?**

→ The maximum profit, not indexes and not the buy/sell prices.

**Step 5 → Do I need indexes?**

→ No, I only need the prices.

**Step 6 → Do I need HashMap or HashSet?**

→ No.

Why?

→ I do not need to remember specific numbers to find later.

→ I only need to remember the **cheapest price so far** and the **best profit so far**.

**Step 7 → What should I remember while moving through prices?**

→ `min_price` = cheapest buying price seen so far.

→ `max_profit` = biggest profit found so far.

**Step 8 → Which loop should I use?**

→ Normal value traversal because I only need the price.

for price in prices:

**Step 9 → When I see a new price, what should I check first?**

→ Is this price cheaper than my current buying price?

if price < min_price:

→ Update `min_price`.

**Step 10 → How do I calculate today's possible profit?**

profit = price - min_price

**Step 11 → Is this new profit better than my previous best profit?**

if profit > max_profit:

→ Update `max_profit`.

**Step 12 → Continue until all prices are checked.**

**Step 13 → Return the biggest profit found.**

---

## 🔍 Recognition

✓ **Array**

→ Input is a list of prices.

✓ **Traversal**

→ I need to check every price.

✓ **Min Tracking**

→ Remember the cheapest price seen so far.

✓ **Max Tracking**

→ Remember the biggest profit seen so far.

✓ **No HashMap**

→ I don't need to store `value → extra information`.

✓ **No HashSet**

→ I don't need to check whether a value was seen before.

✓ **No `enumerate()`**

→ I don't need indexes.

---

# 🧠 When to use HashSet, HashMap, or Neither?

## 1. HashSet → `set()`

Use when you only need to remember:

"Have I seen this value before?"

Example:

Duplicate?

for num in nums:
    if num in seen:
        return True
    seen.add(num)

Mental idea:

Value
↓
Seen before?
↓
Yes / No

Examples:

- Contains Duplicate
- Unique elements
- Seen before?
- Longest substring without repeating characters

---

## 2. HashMap → `{}`

Use when you need to remember:

`value → some information`

Example:

Two Sum:

seen[num] = index

You need:

Number → Index

Other examples:

Character → Frequency

freq[ch] = count

Mental idea:

Key
↓
Extra information

Examples:

- Two Sum → number → index
- Frequency → character → count
- Grouping → key → list
- Previous occurrence → value → position

---

## 3. Neither HashSet nor HashMap

Use neither when the problem only needs you to track a few variables.

Example:

Best Time to Buy and Sell Stock:

min_price
max_profit

You don't need to remember every previous price.

You only care about:

Cheapest price so far
↓
Current possible profit
↓
Best profit so far

Examples:

- Best Time to Buy and Sell Stock
- Maximum Subarray
- Two Pointers
- Binary Search
- Some Min/Max problems

---

## 💻 Code

class Solution(object):
    def maxProfit(self, prices):

        min_price = float('inf')
        max_profit = 0

        for price in prices:

            if price < min_price:
                min_price = price

            profit = price - min_price

            if profit > max_profit:
                max_profit = profit

        return max_profit

## 🔄 Execution

| price | Is price < min_price? | min_price | profit | Is profit > max_profit? | max_profit |
|------:|------------------------|----------:|-------:|--------------------------|-----------:|
| Start | — | ∞ | — | — | 0 |
| 7 | Yes | 7 | 7 - 7 = 0 | 0 > 0 → No | 0 |
| 1 | Yes | 1 | 1 - 1 = 0 | 0 > 0 → No | 0 |
| 5 | No | 1 | 5 - 1 = 4 | 4 > 0 → Yes | 4 |
| 3 | No | 1 | 3 - 1 = 2 | 2 > 4 → No | 4 |
| 6 | No | 1 | 6 - 1 = 5 | 5 > 4 → Yes | 5 |
| 4 | No | 1 | 4 - 1 = 3 | 3 > 5 → No | 5 |

---

## 🧠 Final Logic

New price
↓
Is it the cheapest price so far?
↓
YES → Update `min_price`
↓
Calculate:

profit = current_price - min_price
↓
Is this profit bigger than previous best?
↓
YES → Update `max_profit`
↓
Continue
↓
Return `max_profit`

## 🎯 Quick Recognition Rule

Need only:

"Have I seen this value?"  
→ **HashSet**

Need:

"Value + extra information"  
→ **HashMap**

Need only:

"Track minimum / maximum / current / best"  
→ **No HashMap or HashSet**

````python
def maxProfit(prices):

    min_price = float('inf')
    max_profit = 0
    for price in prices:
        if price < min_price:
            min_price = price
        profit = price - min_price
        if profit > max_profit:
            max_profit = profit
    return max_profit
prices = [7,1,5,3,6,4]
print(maxProfit(prices))
````
